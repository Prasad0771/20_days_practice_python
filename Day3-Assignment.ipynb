Write a function to check whether a string is palindrome or not.

```python
##1.Write a function to check whether a string is palindrome or not.
def is_palindrome(str):
    return str == str[::-1]

txt = input("Enter the text to check for Palindrome")
if is_palindrome(txt):
    print(f"The given string {txt} is a Palindrome")
else:
    print(f"The given string {txt} is not a Palindrome")
        
```

```python
##1.Write a function to check whether a string is palindrome or not.

def check_palindrome(text):
    # Convert string to lowercase
    text = text.lower()

    # Reverse the string
    reversed_text = text[::-1]

    # Check if original and reversed strings are same
    if text == reversed_text:
        print("Palindrome")
    else:
        print("Not a Palindrome")


# User input
word = input("Enter a string: ")

# Function call
check_palindrome(word)
```

Question 2. Write a function to find the second largest number in a list.

```python
def Second_largest(numbers):
    #sorting list in ascending order
    numbers.sort()
    #return 2nd highest number
    return numbers[-2]
# Example list
nums = [10,20,60,40,30,50]
# Function call
print("Second largest number is:", Second_largest(nums))
```

```python
#3.Write a function that accepts a list and returns only unique values.
## The below method preserves order.
def unique_values(lst):
    unique = []
   
    for item in lst:
        if item not in unique:
            unique.append(item)
    return unique

nums = [1,1,7,7,5,6,6,6,8,3,3,4,4]
print(f'The unique values of the list is:, {unique_values(nums)}')
 

    
```

```python
### 8.Use lambda with reduce() to find the product of all numbers in a list
from functools import reduce

num = [1,2,3,4,5,6]
result = reduce(lambda x, y: x * y, num)
type(result)

print(f'The the product of all numbers in a list num = {num}:\n product thr reduce function is :{result}')
```
