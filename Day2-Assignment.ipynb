```python
### Write a program to check whether a year is a leap year.
print("Write a program to check whether a year is a leap year.")
year = int(input("Please Enter Year"))
if year % 4 == 0:
    print("Check if year is century year: ")
    if year % 400 == 0:
        print(f'given year is a leap year: {year}')
    else:
        if year % 100 !=0:
            print(f'given year is a leap year: {year}')
else:
    print(f'given year is not a leap year: {year}')
```

```python
# Q7.Write a program to find the factorial of a number.
num = int(input("enter a number: "))
factorial = 1
for i in range(1, num + 1):
    factorial *= i
print("The factorial of", num, "is", factorial)
```

```python
# Q8.Reverse a given number using loops.
num = int(input("enter a number: "))
print(f'The num : {num} ')
reverse_num = 0
while num >0:
    digit = num%10
    reverse_num = reverse_num*10+digit
    num = num //10
print(f' reverse_num : {reverse_num}')

```

```python
# Q8.Reverse a given number using loops and list
num = int(input("enter a number: "))
print(f'The num : {num} ')
reverse_num = [] 
while num > 0:
    digit = num%10
    reverse_num.append(digit)
    num = num //10
print(f'The reversed number is :{reverse_num}')

```
