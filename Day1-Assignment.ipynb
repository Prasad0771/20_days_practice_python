Q6. Remove duplicates:
x = [1,2,2,3,4,4,5]

```python
### Remove duplicates
x = [1, 2, 2, 3, 4, 4,4,5]
unique_x = set(x)
print(unique_x)
print(type(unique_x))
print('Converting into list')
unique_x = list(set(x))
print(unique_x)
print(type(unique_x))
print('Converting into tuple')
unique_x = tuple(set(x))
print(unique_x)
print(type(unique_x))
```
